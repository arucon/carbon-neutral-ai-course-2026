## 3. 전체 데이터 준비

먼저 패키지에 포함된 **고정 snapshot**을 준비합니다. 이 자료는 출처·행 수·구조와 checksum을 검증한 수업 기준본이므로 인터넷이 연결되지 않아도 12교시를 진행할 수 있습니다.

기본값 `REFRESH_ONLINE = False`는 수업 중 원자료가 갑자기 바뀌는 일을 막습니다. 최신 UCI·OWID 자료를 별도로 확인해야 할 때만 아래 코드의 값을 `True`로 바꾸고, 갱신일과 달라진 결과를 연구기록에 남깁니다.

준비되는 자료는 Bike Sharing, 철강산업 에너지, AI4I 예지보전, OWID CO₂, Kaya-LMDI 및 STIRPAT 실습자료입니다.

아래 표에서 **적재 상태가 `성공`이면 준비가 끝난 것**입니다. `온라인 갱신` 열의 `고정 snapshot 사용`은 다운로드 실패가 아니라 재현 가능한 수업 기준본을 선택했다는 뜻입니다.


```python
from io import BytesIO
import hashlib
import ssl
from urllib.request import Request, urlopen
from zipfile import ZipFile

import certifi
import pandas as pd

DATA_DIR = COURSE_ROOT / "data"
DATA_DIR.mkdir(exist_ok=True)
REFRESH_ONLINE = False

with ZipFile(COURSE_ROOT / "offline_data.zip") as archive:
    archive.extractall(COURSE_ROOT)

print("온라인 갱신:", "사용" if REFRESH_ONLINE else "사용하지 않음 — 고정 snapshot 유지")

def fetch_bytes(url: str) -> bytes:
    request = Request(url, headers={"User-Agent": "Mozilla/5.0 course-data-downloader"})
    contexts = [ssl.create_default_context()]
    certifi_context = ssl.create_default_context(cafile=certifi.where())
    if certifi_context.get_ca_certs() != contexts[0].get_ca_certs():
        contexts.append(certifi_context)

    last_error = None
    for context in contexts:
        try:
            with urlopen(request, timeout=90, context=context) as response:
                return response.read()
        except Exception as exc:
            last_error = exc
    raise last_error

def data_status(destination: Path, expected_rows: int, source: str, online_update: str) -> dict:
    try:
        rows = len(pd.read_csv(destination))
        loaded = destination.exists() and rows == expected_rows
    except Exception:
        rows = 0
        loaded = False
    return {
        "적재 상태": "성공" if loaded else "실패",
        "사용 자료": source,
        "행 수": rows,
        "온라인 갱신": online_update,
    }

def refresh_uci(url: str, member: str, destination: Path, expected_rows: int) -> dict:
    try:
        payload = fetch_bytes(url)
        with ZipFile(BytesIO(payload)) as archive:
            destination.parent.mkdir(parents=True, exist_ok=True)
            destination.write_bytes(archive.read(member))
        return data_status(destination, expected_rows, "공식 공개자료", "완료")
    except Exception as exc:
        return data_status(
            destination,
            expected_rows,
            "배포본 내장 자료",
            f"연결되지 않아 생략 ({type(exc).__name__})",
        )

def snapshot_status(relative_path: str, expected_rows: int) -> dict:
    return data_status(
        DATA_DIR / relative_path,
        expected_rows,
        "배포본 고정 snapshot",
        "고정 snapshot 사용",
    )

status = {}
if REFRESH_ONLINE:
    status["Bike Sharing"] = refresh_uci(
        "https://archive.ics.uci.edu/static/public/275/bike+sharing+dataset.zip",
        "hour.csv",
        DATA_DIR / "bike/hour.csv",
        17_379,
    )
    status["Steel Energy"] = refresh_uci(
        "https://archive.ics.uci.edu/static/public/851/steel+industry+energy+consumption.zip",
        "Steel_industry_data.csv",
        DATA_DIR / "steel/Steel_industry_data.csv",
        35_040,
    )
    status["AI4I"] = refresh_uci(
        "https://archive.ics.uci.edu/static/public/601/ai4i+2020+predictive+maintenance+dataset.zip",
        "ai4i2020.csv",
        DATA_DIR / "ai4i/ai4i2020.csv",
        10_000,
    )
else:
    status["Bike Sharing"] = snapshot_status("bike/hour.csv", 17_379)
    status["Steel Energy"] = snapshot_status("steel/Steel_industry_data.csv", 35_040)
    status["AI4I"] = snapshot_status("ai4i/ai4i2020.csv", 10_000)

if REFRESH_ONLINE:
    try:
        owid_bytes = fetch_bytes("https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv")
        owid = pd.read_csv(
            BytesIO(owid_bytes),
            usecols=["country", "iso_code", "year", "co2", "co2_per_capita"],
        )
        selected = owid.loc[
            owid["country"].isin(["World", "China", "Japan", "South Korea", "United States"])
            & owid["year"].between(1990, 2024)
        ].copy()
        selected["iso_code"] = selected["iso_code"].fillna("OWID_WRL")
        selected = selected.rename(columns={
            "country": "entity",
            "iso_code": "code",
            "co2": "emissions_total",
            "co2_per_capita": "emissions_total_per_capita",
        })
        selected["emissions_total"] = selected["emissions_total"] * 1_000_000
        selected = selected[["entity", "code", "year", "emissions_total", "emissions_total_per_capita"]]
        if len(selected) != 175 or selected[["code", "year"]].duplicated().any():
            raise ValueError(f"unexpected OWID selection: {selected.shape}")
        destination = DATA_DIR / "co2/owid_co2_sample.csv"
        destination.parent.mkdir(parents=True, exist_ok=True)
        selected.to_csv(destination, index=False)
        status["OWID CO2"] = data_status(destination, 175, "공식 공개자료", "완료")
    except Exception as exc:
        status["OWID CO2"] = data_status(
            DATA_DIR / "co2/owid_co2_sample.csv",
            175,
            "배포본 고정 snapshot",
            f"연결되지 않아 snapshot 유지 ({type(exc).__name__})",
        )
else:
    status["OWID CO2"] = snapshot_status("co2/owid_co2_sample.csv", 175)

def file_sha256(path: Path) -> str:
    return hashlib.sha256(path.read_bytes()).hexdigest()

kaya_path = DATA_DIR / "methods/kaya_lmdi_sample.csv"
stirpat_path = DATA_DIR / "methods/stirpat_panel.csv"
kaya = pd.read_csv(kaya_path)
stirpat = pd.read_csv(stirpat_path)
kaya_columns = {
    "entity", "code", "year", "co2_tonnes", "co2_tonnes_per_capita",
    "gdp_per_capita_intl2021", "primary_energy_twh",
}
kaya_coverage_ok = (
    set(kaya["code"]) == {"DEU", "JPN", "KOR", "USA"}
    and kaya["year"].min() == 1990
    and kaya["year"].max() == 2024
    and kaya.groupby("code")["year"].count().eq(35).all()
    and not kaya.duplicated(["code", "year"]).any()
)
method_ok = (
    len(kaya) == 140
    and set(kaya.columns) == kaya_columns
    and kaya_coverage_ok
    and file_sha256(kaya_path) == "e1de73357bba5340342c51d33ec2040db2fdd6902c2a36cf52894b57805f105f"
    and len(stirpat) == 396
    and file_sha256(stirpat_path) == "8dce11a56dc1add833bd764fb3fc10758f7791eae4e15ec728d9e6157e626b97"
)
status["Kaya-LMDI / STIRPAT"] = {
    "적재 상태": "성공" if method_ok else "실패",
    "사용 자료": "출처·구조·checksum 검증 snapshot",
    "행 수": f"{len(kaya)} / {len(stirpat)}",
    "온라인 갱신": "고정 snapshot 사용",
}

status_table = pd.DataFrame.from_dict(status, orient="index")
display(status_table)
if status_table["적재 상태"].eq("성공").all():
    print("DATA LOAD: PASS — 모든 수업 데이터가 준비되었습니다.")
else:
    raise RuntimeError("적재 상태가 '실패'인 행을 확인해 주세요.")
```
