```python
from pathlib import Path
import platform
import warnings

import numpy as np
import pandas as pd
from IPython.display import display

pd.set_option("display.max_columns", 40)
pd.set_option("display.width", 130)
warnings.filterwarnings("ignore", category=FutureWarning)

def find_course_data(relative_path: str) -> Path:
    """0교시에서 준비한 data 폴더의 파일을 찾습니다."""
    relative = Path(relative_path)
    checked = []
    for start in [Path.cwd(), *Path.cwd().parents, Path("/content")]:
        candidate = (start / "data" / relative).resolve()
        if candidate in checked:
            continue
        checked.append(candidate)
        if candidate.exists():
            return candidate
    raise FileNotFoundError(
        f"데이터 파일을 찾지 못했습니다: {relative_path}. "
        "과정 폴더의 00_setup.ipynb를 먼저 실행해 주세요."
    )
print("Python:", platform.python_version())
print("pandas:", pd.__version__)
```

## 1. 경로와 입력정보 조건 선택

공통 Bike 경로는 끝까지 실행 가능한 기준 분석입니다. `BIKE_VARIANT`의 두 값은 실제 입력변수 목록을 바꿉니다.

| 선택값 | 입력정보 | 해석 범위 |
|---|---|---|
| `calendar_and_lag` | 직전 수요와 사전에 아는 달력정보 | 운영 전 예측에 가까운 기본 경로 |
| `observed_weather_benchmark` | 위 정보와 결과시각의 실제 날씨 관측 | 실제 예보성능이 아닌 조건부 benchmark |

개인·조별 경로는 아래 `PERSONAL_PROJECT`의 빈칸을 채워 최소 분석계획을 검토합니다. 원자료 연결은 자료 사용권과 스키마를 확인한 뒤 별도로 진행합니다.


```python
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, root_mean_squared_error
import json

PROJECT_TRACK = "bike"  # 허용값: "bike", "personal_canvas"
BIKE_VARIANT = "calendar_and_lag"  # 기본값. 다른 허용값: "observed_weather_benchmark"

# Bike 실습에서 자신의 판단으로 고쳐 쓰는 다섯 필드입니다.
MY_EXPECTATION = "달력과 직전 수요만 사용하면 실제 운영 전 정보경계가 더 분명해질 것으로 예상합니다."
WHY = "결과시각의 실제 날씨는 예측을 만드는 시점에 아직 관측되지 않았기 때문입니다."
ONE_CHANGE = "입력조건을 calendar_and_lag로 정하고 결과시각의 실측 날씨 열을 제외했습니다."
RISK = "단일 도시와 한 번의 미래기간만 평가해 계절별 안정성을 알 수 없습니다."
REVISION_DECISION = "다음 분석에서는 예측 기준시점을 옮겨 여러 계절을 반복 평가합니다."

ALLOWED_TRACKS = {"bike", "personal_canvas"}
ALLOWED_BIKE_VARIANTS = {"calendar_and_lag", "observed_weather_benchmark"}
if PROJECT_TRACK not in ALLOWED_TRACKS:
    raise ValueError(f"PROJECT_TRACK은 {sorted(ALLOWED_TRACKS)} 중 하나여야 합니다.")
if BIKE_VARIANT not in ALLOWED_BIKE_VARIANTS:
    raise ValueError(f"BIKE_VARIANT는 {sorted(ALLOWED_BIKE_VARIANTS)} 중 하나여야 합니다.")

RESEARCH_QUESTION = (
    "과거 수요와 예측시점에 사용할 수 있다고 정한 정보로 미래 시간당 Bike 수요를 예측할 때, "
    "후보모형의 미래 test MAE가 직전 1시간 수요 기준모형보다 낮은가?"
)
bike_canvas = pd.DataFrame({
    "항목": ["현실 문제", "분석 질문", "관측 단위", "결과변수", "예측 기준시점",
             "비교 기준모형", "자료 분할", "평가척도", "결과 사용자"],
    "기록": ["시간대별 자전거 배치계획", RESEARCH_QUESTION, "한 도시의 한 시간",
             "cnt: 시간당 대여건수", "결과시각 1시간 전", "직전 1시간 수요",
             "과거 80% 학습 → 미래 20% 평가", "MAE·RMSE (건/시간)", "운영계획 담당자"],
})

PERSONAL_PROJECT = {
    "현실 문제": "",
    "분석 질문": "",
    "관측 단위": "",
    "결과변수": "",
    "정보 기준시점": "",
    "비교 기준": "",
    "분할 또는 집단비교": "",
    "평가척도와 단위": "",
    "자료 출처·사용권": "",
    "다음 실행": "",
}
personal_canvas = pd.DataFrame({"항목": list(PERSONAL_PROJECT), "기록": list(PERSONAL_PROJECT.values())})
active_canvas = bike_canvas if PROJECT_TRACK == "bike" else personal_canvas

print("선택한 경로:", PROJECT_TRACK)
if PROJECT_TRACK == "bike":
    print("Bike 입력조건:", BIKE_VARIANT)
    display(bike_canvas)
else:
    print("개인·조별 분석계획 공유 경로입니다. 이 경로에서는 Bike 모델·성능수치를 개인 기록에 넣지 않습니다.")
    display(personal_canvas)
```

## 2. 한 행·정보시점·평가기간 고정

정확한 1시간 lag를 timestamp로 연결하고 앞 80%를 학습, 뒤 20%를 평가에 사용합니다. 선택한 입력정보가 결과시각에 언제 확보되는지 표로 확인합니다.


```python
feature_options = {
    "calendar_and_lag": ["lag_1h", "hr", "weekday", "workingday"],
    "observed_weather_benchmark": [
        "lag_1h", "hr", "weekday", "workingday", "weathersit", "temp", "hum", "windspeed"
    ],
}
TARGET_TIME_WEATHER_COLUMNS = ["weathersit", "temp", "hum", "windspeed"]
WEATHER_BENCHMARK_NOTE = '이 결과는 결과시각의 실제 날씨 관측값을 사용할 수 있다고 가정한 조건부 benchmark입니다. 실제 운영 예측에는 예보값과 예보오차 검증이 필요합니다.'

if PROJECT_TRACK == "bike":
    data_path = find_course_data("bike/hour.csv")
    raw = pd.read_csv(data_path, parse_dates=["dteday"])
    raw["timestamp"] = raw["dteday"] + pd.to_timedelta(raw["hr"], unit="h")
    raw = raw.sort_values("timestamp")
    lag = raw[["timestamp", "cnt"]].copy()
    lag["timestamp"] += pd.Timedelta(hours=1)
    data = raw.merge(lag.rename(columns={"cnt": "lag_1h"}), on="timestamp", how="left")
    data = data.dropna(subset=["lag_1h"]).copy()
    features = feature_options[BIKE_VARIANT]
    TARGET_TIME_WEATHER_USED = sorted(set(features) & set(TARGET_TIME_WEATHER_COLUMNS))
    cut = int(len(data) * 0.80)
    train, test = data.iloc[:cut], data.iloc[cut:]
    feature_availability = pd.DataFrame({
        "입력": ["lag_1h", "hr·weekday·workingday", "weathersit·temp·hum·windspeed"],
        "값의 기준시각": ["1시간 전", "결과시각", "결과시각"],
        "예측 때 확보": ["직전 측정이 즉시 들어온다고 가정", "달력으로 사전 확인", "자료에는 실측값만 있음"],
        "현재 선택": [True, True, bool(TARGET_TIME_WEATHER_USED)],
        "판정": ["가정 아래 사용", "사용 가능", "사용 시 예보자료 필요"],
    })
    print("train/test dates:", train["timestamp"].min(), train["timestamp"].max(), "|", test["timestamp"].min(), test["timestamp"].max())
    print("선택한 입력변수:", features)
    display(feature_availability)
    display(data[["timestamp", "cnt", *features]].head(5))
    print(WEATHER_BENCHMARK_NOTE if TARGET_TIME_WEATHER_USED else "결과시각의 실제 날씨 관측값을 입력에서 제외했습니다.")
else:
    data_path = None
    data = train = test = None
    features = []
    TARGET_TIME_WEATHER_USED = []
    print("개인·조별 경로는 분석계획만 검토합니다. 원자료·Bike 결과·모형은 실행하지 않습니다.")
```

## 3. 같은 미래기간에서 기준모형과 후보모형 비교

직전 1시간 수요를 반복하는 persistence와 Gradient Boosting을 같은 평가행의 MAE·RMSE로 비교합니다. 그림에는 실제값과 두 예측을 모두 표시합니다.


```python
if PROJECT_TRACK == "bike":
    model = HistGradientBoostingRegressor(max_iter=120, learning_rate=0.07, random_state=42)
    model.fit(train[features], train["cnt"])
    prediction = model.predict(test[features])
    baseline = test["lag_1h"].to_numpy()
    result_table = pd.DataFrame([
        {"model": "persistence", "MAE": mean_absolute_error(test["cnt"], baseline),
         "RMSE": root_mean_squared_error(test["cnt"], baseline)},
        {"model": "hist gradient boosting", "MAE": mean_absolute_error(test["cnt"], prediction),
         "RMSE": root_mean_squared_error(test["cnt"], prediction)},
    ])
    display(result_table)
    result_by_model = result_table.set_index("model")
    candidate_mae = float(result_by_model.loc["hist gradient boosting", "MAE"])
    baseline_mae = float(result_by_model.loc["persistence", "MAE"])
    mae_change = baseline_mae - candidate_mae
    mae_direction = "낮았습니다" if mae_change >= 0 else "높았습니다"
    evaluation_period = f"{test['timestamp'].min():%Y-%m-%d %H:%M}–{test['timestamp'].max():%Y-%m-%d %H:%M}"

    import matplotlib.pyplot as plt
    view = 168
    fig, ax = plt.subplots(figsize=(10.5, 4.3))
    ax.plot(test["timestamp"].iloc[:view], test["cnt"].iloc[:view], label="observed")
    ax.plot(test["timestamp"].iloc[:view], baseline[:view], label="persistence baseline", alpha=0.85)
    ax.plot(test["timestamp"].iloc[:view], prediction[:view], label="candidate")
    ax.set(title=f"Research sprint | {BIKE_VARIANT} | first test week", xlabel="Time", ylabel="Rentals/hour")
    ax.legend(); plt.xticks(rotation=25); plt.tight_layout(); plt.show()
    result_claim = (
        f"{evaluation_period}의 시간순 test에서 후보모형 MAE는 {candidate_mae:.2f}건/시간으로, "
        f"persistence의 {baseline_mae:.2f}건/시간보다 {abs(mae_change):.2f}건/시간 {mae_direction}."
    )
    print(result_claim)
else:
    model = prediction = baseline = result_table = None
    candidate_mae = baseline_mae = mae_change = mae_direction = None
    evaluation_period = None
    result_claim = None
    print("개인·조별 분석계획 공유 경로: 결과수치·모형·성능표를 생성하지 않습니다.")
```

## 4. 수정기록과 60초 공유문

선택한 경로·입력조건, 변경 전 예상, 현재 확보한 근거, 가장 큰 위험과 다음 검증을 한 기록으로 묶습니다. Bike 공유문에는 대표 수치를 넣고, 개인·조별 공유문에는 아직 결과가 없다는 사실과 다음 실행을 넣습니다. 두 경로 모두 같은 artifact 파일에 서로 다른 `status`로 저장합니다.


```python
PLACEHOLDER_MARKERS = {"", "TODO", "TBD", "여기에 작성", "작성 필요"}

def is_filled(value):
    return str(value).strip() not in PLACEHOLDER_MARKERS and len(str(value).strip()) >= 8

artifact_dir = Path.cwd() / "artifacts"
artifact_dir.mkdir(parents=True, exist_ok=True)
artifact_path = artifact_dir / "lesson11_project_record.json"

if PROJECT_TRACK == "bike":
    limitation = (
        WEATHER_BENCHMARK_NOTE if TARGET_TIME_WEATHER_USED else
        "단일 도시·기간·분할 결과이며 다른 연도와 도시의 성능은 검증하지 않았습니다."
    )
    editable_fields = {
        "MY_EXPECTATION": MY_EXPECTATION,
        "WHY": WHY,
        "ONE_CHANGE": ONE_CHANGE,
        "RISK": RISK,
        "REVISION_DECISION": REVISION_DECISION,
    }
    modification_record = pd.Series(editable_fields, name="수정기록")
    display(modification_record.to_frame())
    share_question = (
        "직전 수요와 예측 전에 확정되는 달력정보로 미래 한 시간 수요를 예측할 때 후보모형이 직전 값 기준보다 나은지"
        if not TARGET_TIME_WEATHER_USED else
        "결과시각의 실측 날씨가 주어졌다고 가정한 조건부 benchmark에서 후보모형이 직전 값 기준보다 나은지"
    )
    share_60s = (
        f"질문은 {share_question}입니다. "
        f"한 행은 한 도시의 한 시간이며 {evaluation_period}을 평가했습니다. "
        f"입력조건은 {BIKE_VARIANT}이고, persistence MAE {baseline_mae:.2f}와 후보 MAE {candidate_mae:.2f}건/시간을 비교했습니다. "
        f"차이는 {abs(mae_change):.2f}건/시간 {mae_direction}. 가장 큰 위험은 {RISK} 수정 결정은 {REVISION_DECISION}"
    )
    artifact_payload = {
        "schema_version": 1,
        "track": "bike",
        "status": "analysis_result",
        "bike_variant": BIKE_VARIANT,
        "canvas": bike_canvas.to_dict("records"),
        "data_relative_path": "bike/hour.csv",
        "features": features,
        "target_time_observed_weather": TARGET_TIME_WEATHER_USED,
        "train_end": str(train["timestamp"].max()),
        "test_start": str(test["timestamp"].min()),
        "evaluation_period": evaluation_period,
        "model": "HistGradientBoostingRegressor(max_iter=120, learning_rate=0.07, random_state=42)",
        "metrics": result_table.to_dict("records"),
        "result_claim": result_claim,
        "limitation": limitation,
        "reflection": editable_fields,
        "share_type": "60-second analysis result",
        "share_text": share_60s,
    }
    checks = pd.Series({
        "valid Bike variant": BIKE_VARIANT in ALLOWED_BIKE_VARIANTS,
        "all five learner fields completed": all(is_filled(value) for value in editable_fields.values()),
        "strict time order": train["timestamp"].max() < test["timestamp"].min(),
        "target and leakage columns excluded": not ({"cnt", "casual", "registered"} & set(features)),
        "weather condition preserved": bool(TARGET_TIME_WEATHER_USED) == (BIKE_VARIANT == "observed_weather_benchmark"),
        "baseline and candidate": set(result_table["model"]) == {"persistence", "hist gradient boosting"},
        "result claim has period and both MAE": evaluation_period in result_claim and all(f"{v:.2f}" in result_claim for v in [candidate_mae, baseline_mae]),
        "60-second result share has evidence": all(token in share_60s for token in ["MAE", f"{candidate_mae:.2f}", "위험"]),
    }, name="passed")
else:
    personal_canvas_complete = personal_canvas["기록"].astype(str).map(is_filled).all()
    planning_share = (
        f"저희는 {PERSONAL_PROJECT['현실 문제']}를 다루기 위한 분석계획을 공유합니다. "
        f"질문은 {PERSONAL_PROJECT['분석 질문']}이며 한 행은 {PERSONAL_PROJECT['관측 단위']}입니다. "
        f"결과변수는 {PERSONAL_PROJECT['결과변수']}, 비교 기준은 {PERSONAL_PROJECT['비교 기준']}, "
        f"평가척도는 {PERSONAL_PROJECT['평가척도와 단위']}입니다. 아직 결과를 계산하지 않았으며 다음 실행은 {PERSONAL_PROJECT['다음 실행']}입니다."
    )
    share_60s = planning_share
    limitation = "현재 경로는 승인 전 분석계획이며 모형·성능수치·결과주장을 포함하지 않습니다."
    artifact_payload = {
        "schema_version": 1,
        "track": "personal_canvas",
        "status": "analysis_plan",
        "canvas": personal_canvas.to_dict("records"),
        "data_source_and_license": PERSONAL_PROJECT["자료 출처·사용권"],
        "information_timing": PERSONAL_PROJECT["정보 기준시점"],
        "baseline": PERSONAL_PROJECT["비교 기준"],
        "metric": PERSONAL_PROJECT["평가척도와 단위"],
        "next_execution": PERSONAL_PROJECT["다음 실행"],
        "model": None,
        "metrics": None,
        "result_claim": None,
        "limitation": limitation,
        "share_type": "60-second analysis plan",
        "share_text": planning_share,
    }
    checks = pd.Series({
        "personal canvas complete": personal_canvas_complete,
        "source and license recorded": is_filled(PERSONAL_PROJECT["자료 출처·사용권"]),
        "information timing recorded": is_filled(PERSONAL_PROJECT["정보 기준시점"]),
        "baseline and metric recorded": is_filled(PERSONAL_PROJECT["비교 기준"]) and is_filled(PERSONAL_PROJECT["평가척도와 단위"]),
        "next execution recorded": is_filled(PERSONAL_PROJECT["다음 실행"]),
        "no Bike metrics or model mixed in": artifact_payload["metrics"] is None and artifact_payload["model"] is None,
        "planning share makes no result claim": "아직 결과를 계산하지 않았" in planning_share,
    }, name="passed")

artifact_path.write_text(json.dumps(artifact_payload, ensure_ascii=False, indent=2), encoding="utf-8")
print("\n[60초 공유문]")
print(share_60s)
print("\nL11 artifact:", artifact_path)
display(checks.to_frame()); assert checks.all()
print("연구 분석흐름 검증: PASS")

review_prompt = f"""
아래 공개 가능한 {artifact_payload['share_type']} 기록만 검토해 주세요.

{json.dumps(artifact_payload, ensure_ascii=False, indent=2)}

1. 판정: 통과 / 수정 필요 / 정보 부족
2. 가장 큰 설계 위험 1개와 기록 안의 근거
3. 다음 실행 전에 확인할 한 가지

분석계획 경로에는 결과수치나 모형성능을 만들어 넣지 마세요. 제공되지 않은 정보는 추정하지 마세요.
""".strip()
print("\n[ChatGPT에 복사할 프롬프트]\n")
print(review_prompt)
```
