## 근거·주장·한계의 연결

| 구성 | Bike 예측결과의 예 |
|---|---|
| 근거(evidence) | 미래 test에서 boosting MAE가 persistence보다 낮음 |
| 주장(claim) | 선택한 미래 test 기간에서 후보모형의 예측오차가 baseline보다 작았음 |
| 한계(limitation) | 단일 도시·기간·split이며 외부연도와 실제 날씨예보 오차는 검증하지 않음 |

예측 성능은 선택한 미관측 표본에서의 오차 차이를 보여줍니다. 이 근거만으로 정책 개입의 효과나 다른 도시에서의 성능을 주장할 수는 없습니다.

## 분석방법에 맞는 결과 표현

| 분석 결과 | 적절한 보고 표현 | 해석범위를 벗어난 표현 |
|---|---|---|
| holdout 예측 | 선택한 test에서 오차가 낮았다 | 결과의 원인을 개선했다 |
| 회귀계수 | 선택한 명세에서 양의 연관성을 보였다 | 결과에 영향을 주었다 |
| LMDI 분해 | 총변화가 증가·감소 방향의 기여로 분해되었다 | 정책의 인과효과였다 |
| 인과분석 | 식별가정 아래 개입효과를 추정했다 | 가정과 설계 없이 원인이라고 단정했다 |

결과문장에는 표본, 기간, 명세, 평가척도를 포함하고 한계문장에는 현재 설계가 검증하지 않은 가장 가까운 범위를 적습니다.

## 1. 새 실행환경에서 분석 재현

1. Notebook을 저장하고 kernel을 재시작합니다.
2. 데이터 읽기부터 모든 셀을 순서대로 실행합니다.
3. 행 수, 기간, 파일경로, 라이브러리 버전을 기록합니다.
4. 입력·결과변수, split, seed, 모형설정을 출력합니다.
5. 결과표와 검증값이 동일하게 만들어지는지 확인합니다.

재실행 실패는 숨은 변수나 셀 순서 의존성을 의미할 수 있습니다. 오류가 발생한 셀과 마지막 정상 출력을 기록한 뒤 보고 전에 수정합니다.


```python
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, root_mean_squared_error
import sklearn

analysis_name = "reproducible Bike report"
data_path = find_course_data("bike/hour.csv")
raw = pd.read_csv(data_path, parse_dates=["dteday"])
raw["timestamp"] = raw["dteday"] + pd.to_timedelta(raw["hr"], unit="h")
raw = raw.sort_values("timestamp")
lag = raw[["timestamp", "cnt"]].copy(); lag["timestamp"] += pd.Timedelta(hours=1)
data = raw.merge(lag.rename(columns={"cnt":"lag_1h"}), on="timestamp", how="left").dropna(subset=["lag_1h"]).copy()
features = ["lag_1h", "hr", "weekday", "workingday", "weathersit", "temp", "hum", "windspeed"]
TARGET_TIME_WEATHER_COLUMNS = ["weathersit", "temp", "hum", "windspeed"]
TARGET_TIME_WEATHER_USED = sorted(set(features) & set(TARGET_TIME_WEATHER_COLUMNS))
WEATHER_BENCHMARK_NOTE = '이 결과는 결과시각의 실제 날씨 관측값을 사용할 수 있다고 가정한 조건부 benchmark입니다. 실제 운영 예측에는 예보값과 예보오차 검증이 필요합니다.'
cut = int(len(data)*.8); train, test = data.iloc[:cut], data.iloc[cut:]
model = HistGradientBoostingRegressor(max_iter=120, learning_rate=.07, random_state=42).fit(train[features], train["cnt"])
prediction = model.predict(test[features]); baseline = test["lag_1h"].to_numpy()
```

### 출력 해석: 재실행 직후 확보할 정보

데이터 범위·분할 경계·feature·첫 예측을 한 화면에서 확인합니다.


```python
print("rows after lag join:", len(data))
print("train:", train["timestamp"].min(), "→", train["timestamp"].max())
print("test :", test["timestamp"].min(), "→", test["timestamp"].max())
print("features:", features)
report_preview = pd.DataFrame({
    "timestamp": test["timestamp"].iloc[:8].to_numpy(),
    "observed": test["cnt"].iloc[:8].to_numpy(),
    "baseline": baseline[:8],
    "candidate": prediction[:8],
})
display(report_preview)
```

결과표는 전체 test 성능을 요약하고 그림은 선택한 구간의 패턴을 보여줍니다. 두 자료의 범위가 다르면 제목과 caption에 각각의 범위를 적습니다.

> **확인 질문:** 제목과 caption만으로 평가기간, 비교대상, 단위를 확인할 수 있습니까?


```python
result_table = pd.DataFrame([
    {"model":"previous-hour persistence", "MAE_rentals_per_hour":mean_absolute_error(test["cnt"], baseline), "RMSE_rentals_per_hour":root_mean_squared_error(test["cnt"], baseline)},
    {"model":"hist gradient boosting", "MAE_rentals_per_hour":mean_absolute_error(test["cnt"], prediction), "RMSE_rentals_per_hour":root_mean_squared_error(test["cnt"], prediction)},
])
display(result_table)
result_by_model = result_table.set_index("model")
mae_difference = (
    result_by_model.loc["previous-hour persistence", "MAE_rentals_per_hour"]
    - result_by_model.loc["hist gradient boosting", "MAE_rentals_per_hour"]
)
mae_difference_direction = "낮습니다" if mae_difference >= 0 else "높습니다"
print(f"MAE difference | 후보모형의 MAE는 persistence보다 {abs(mae_difference):.2f}건/시간 {mae_difference_direction}.")

import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(10.5,4.3)); view=168
ax.plot(test["timestamp"].iloc[:view], test["cnt"].iloc[:view], label="observed")
ax.plot(test["timestamp"].iloc[:view], baseline[:view], label="baseline", alpha=0.85)
ax.plot(test["timestamp"].iloc[:view], prediction[:view], label="candidate")
figure_start = test["timestamp"].iloc[0]
figure_end = test["timestamp"].iloc[view - 1]
ax.set(title=f"Figure 1. Test {figure_start:%Y-%m-%d}–{figure_end:%Y-%m-%d}: first 168 hours", xlabel="Timestamp", ylabel="Bike rentals/hour")
ax.legend(); plt.xticks(rotation=25); plt.tight_layout(); plt.show()
print("\n[입력정보 해석 조건]", WEATHER_BENCHMARK_NOTE)
```

## 3. 근거 범위에 맞는 결과문장

| 분석 결과 | 적절한 표현 | 해석범위를 벗어난 표현 |
|---|---|---|
| holdout 예측 | 선택한 test에서 오차가 낮았다 | 원인을 개선했다 |
| 회귀계수 | 선택한 명세에서 연관성을 보였다 | 영향을 주었다 |
| LMDI 기여도 | 총변화가 방향별 기여로 분해되었다 | 정책효과였다 |

결과문장에는 표본, 기간, 분석명세, metric을 포함합니다. 한계문장에는 단일 기간, 외부집단 미검증, 입력정보의 차이처럼 현재 주장을 직접 제한하는 조건을 기록합니다.

> **확인 질문:** 결과표가 직접 지지하지 않는 동사가 문장에 포함되어 있습니까?


```python
candidate_mae = result_by_model.loc["hist gradient boosting", "MAE_rentals_per_hour"]
baseline_mae = result_by_model.loc["previous-hour persistence", "MAE_rentals_per_hour"]
mae_direction = "낮았다" if mae_difference >= 0 else "높았다"
evaluation_period = f"{test['timestamp'].min():%Y-%m-%d %H:%M}–{test['timestamp'].max():%Y-%m-%d %H:%M}"
claim = (
    f"{evaluation_period}의 시간순 평가자료에서 HistGradientBoosting의 MAE는 "
    f"{candidate_mae:.2f}건/시간으로, previous-hour persistence의 "
    f"{baseline_mae:.2f}건/시간보다 {abs(mae_difference):.2f}건/시간 {mae_direction}."
)
limitation = "단일 도시·기간·split 결과입니다. " + WEATHER_BENCHMARK_NOTE
reproducibility_record = pd.Series({
    "source": "UCI Bike Sharing, DOI 10.24432/C5W894",
    "data_snapshot": f"course offline snapshot: {data_path.name}",
    "evaluation_period": evaluation_period,
    "rows_after_lag_join": len(data),
    "train_end": str(train["timestamp"].max()),
    "test_start": str(test["timestamp"].min()),
    "features": features,
    "model": "HistGradientBoostingRegressor(max_iter=120, learning_rate=0.07, random_state=42)",
    "python": platform.python_version(), "pandas": pd.__version__, "scikit_learn": sklearn.__version__,
    "claim_type": "prediction performance on held-out future period",
    "target_time_observed_weather": TARGET_TIME_WEATHER_USED,
    "weather_scope_note": WEATHER_BENCHMARK_NOTE,
    "limitation": limitation,
}, name="record")
print("CLAIM:", claim); print("LIMITATION:", limitation)
display(reproducibility_record.to_frame())
```

### 출력 해석: 결과 문장과 한계 문장

성능 숫자, 주장 범위, 한계가 서로 다른 역할을 갖는다는 점을 표로 구분합니다.


```python
claim_view = pd.DataFrame({
    "record type": ["result", "claim", "limitation"],
    "content": [
        result_table.round(3).to_string(index=False),
        claim,
        limitation,
    ],
})
display(claim_view)
```

## 4. 3분 발표 점검표

1. 현실 문제와 분석질문을 30초 안에 설명합니다.
2. 관측 단위, 기간, 방법, baseline을 제시합니다.
3. 표 또는 그림에서 핵심 수치를 직접 가리킵니다.
4. 결과문장과 가장 큰 한계를 구분합니다.
5. 다음 검증과 청중에게 받을 질문을 제시합니다.

발표의 구성은 `문제 → 방법 → 근거 → 주장과 한계 → 다음 검증`입니다. 각 항목이 같은 데이터와 평가조건을 가리키는지 마지막으로 확인합니다.


```python
problem_text = "시간대별 자전거 배치계획을 위해 다음 한 시간의 대여량을 예측했습니다."
method_text = (
    f"한 행은 한 도시의 한 시간이며 {train['timestamp'].min():%Y-%m-%d}부터 "
    f"{train['timestamp'].max():%Y-%m-%d}까지 학습하고 이후 기간을 평가했습니다. "
    "직전 1시간 값 기준모형과 HistGradientBoosting을 비교했습니다."
)
result_text = claim
limitation_text = limitation
next_check_text = "실제 날씨예보 자료와 여러 예측 기준시점에서 성능을 다시 확인하겠습니다."
presentation = pd.DataFrame({
    "시간": ["0:00–0:30", "0:30–1:00", "1:00–2:00", "2:00–2:30", "2:30–3:00"],
    "실제로 말할 내용": [problem_text, method_text, result_text, limitation_text, next_check_text],
})
display(presentation)
checks = pd.Series({
    "two comparable results": len(result_table) == 2,
    "finite metrics": np.isfinite(result_table.select_dtypes("number")).all().all(),
    "claim has period": evaluation_period in claim,
    "claim has both MAE values": all(f"{value:.2f}" in claim for value in [candidate_mae, baseline_mae]),
    "claim has difference and unit": f"{abs(mae_difference):.2f}" in claim and "건/시간" in claim,
    "weather scope recorded": "조건부 benchmark" in limitation,
    "presentation five parts complete": len(presentation) == 5 and presentation["실제로 말할 내용"].str.strip().ne("").all(),
}, name="passed")
display(checks.to_frame()); assert checks.all()
print("재현성과 보고조건 검증: PASS")
```

## ChatGPT 저장 응답 검증: 결과문장이 증거 범위를 넘는지 확인하기

최종 표·그림의 수치, 결과문장과 한계문장을 함께 제공해 서로 같은 분석을 가리키는지 점검합니다. ChatGPT에는 새 결과를 만들게 하지 않고 **과장된 표현의 최소 수정과 누락된 재현정보**만 요청합니다.

채택한 문장은 원 결과에서 숫자를 다시 찾아 확인하고, 생성형 AI 사용 범위와 연구자가 직접 검증한 내용을 함께 기록합니다.


```python
review_prompt = f"""
다음 연구결과의 증거–주장–한계 연결을 검토해 주세요.

- 결과표:\n{result_table.to_string(index=False)}
- 결과문장: {claim}
- 한계문장: {limitation}
- 재현성 기록:\n{reproducibility_record.to_string()}

아래 형식으로 답해 주세요.
1. 결과문장의 유형: 예측 / 연관성 / 인과
2. 결과표보다 강한 표현과 최소 수정안
3. 빠진 표본·기간·단위·비교조건·재현정보
4. 30초 청중 질문 1개와 근거에 기반한 답변
5. 생성형 AI가 도운 범위와 사람이 검증한 범위를 밝히는 문장

제약: 새로운 성능 숫자나 원인을 만들지 마세요.
""".strip()
print("[ChatGPT에 복사할 프롬프트]\n")
print(review_prompt)
```

### ChatGPT 저장 응답을 검증할 때

1. 문장에 들어간 숫자를 결과표에서 하나씩 다시 찾습니다.
2. 표본·기간·비교대상·단위와 한계가 빠지지 않았는지 확인합니다.
3. 최종 보고에는 생성형 AI가 도운 범위와 사람이 검증한 범위를 구분해 적습니다.
