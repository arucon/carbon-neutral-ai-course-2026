# Day3_02 · 원하는 데이터 찾기

측정소 이름과 숫자 조건을 이용해 필요한 행만 찾습니다.

## 이 노트북에서 확인할 내용

- 서로 다른 값을 확인할 수 있습니다.
- 하나의 조건과 두 개의 조건으로 행을 찾을 수 있습니다.
- 조건값을 바꾸고 결과 행 수를 비교할 수 있습니다.

## 실행 방법

위에서 아래로 코드 셀을 실행하고 바로 아래 결과를 확인합니다. 값을 만드는 셀 다음에는 그 값을 화면에 표시하는 셀이 이어집니다. 수정 안내가 있는 셀에서는 주석 아래의 값만 바꿉니다.

### D3B-02-01 · 경로 도구 불러오기

```python
from pathlib import Path
```

### D3B-02-02 · 현재 폴더 확인

```python
current_folder = Path.cwd()
```

```python
current_folder
```

### D3B-02-03 · 데이터 폴더 후보 만들기

```python
data_candidates = [
    current_folder / 'data',
    current_folder.parent / 'data',
]
```

```python
data_candidates
```

### D3B-02-04 · 데이터 폴더 선택

```python
DATA_DIR = next(folder for folder in data_candidates if folder.exists())
```

```python
DATA_DIR
```

### D3B-02-05 · pandas 불러오기

```python
import pandas as pd
```

### D3B-02-06 · CSV 읽기

```python
air = pd.read_csv(DATA_DIR / 'air_quality.csv')
```

```python
air
```

### D3B-02-07 · 측정소 이름 모으기

```python
station_names = air['station'].unique()
```

```python
station_names
```

### D3B-02-08 · 확인할 측정소 정하기

```python
# River-1을 Central-1 또는 Park-1로 바꿔 보세요.
selected_station = 'River-1'
```

```python
selected_station
```

### D3B-02-09 · 한 측정소의 행 찾기

```python
station_rows = air[air['station'] == selected_station]
```

```python
station_rows
```

### D3B-02-10 · 찾은 행 수 확인

```python
station_rows.shape
```

### D3B-02-11 · PM2.5 기준 정하기

```python
# 아래 숫자를 30 또는 40으로 바꿔 보세요.
pm25_limit = 35
```

```python
pm25_limit
```

### D3B-02-12 · 기준 이상인 행 찾기

```python
high_pm25 = air[air['pm25'] >= pm25_limit]
```

```python
high_pm25
```

### D3B-02-13 · 기준 이상인 행 수 확인

```python
high_pm25.shape
```

### D3B-02-14 · 필요한 열만 확인

```python
high_pm25[['date', 'station', 'pm25']].head(10)
```

### D3B-02-15 · 두 조건을 동시에 적용

```python
station_and_high = air[(air['station'] == selected_station) & (air['pm25'] >= pm25_limit)]
```

```python
station_and_high
```

### D3B-02-16 · 기온 기준 정하기

```python
# 아래 숫자를 8 또는 12로 바꿔 보세요.
temperature_limit = 10
```

```python
temperature_limit
```

### D3B-02-17 · 기온과 PM2.5 조건 함께 적용

```python
cold_and_high = air[(air['temperature'] < temperature_limit) & (air['pm25'] >= pm25_limit)]
```

```python
cold_and_high
```

### D3B-02-18 · 날짜순으로 정렬

```python
cold_and_high.sort_values('date').head(10)
```

## 반복 실습 · 조건이 작동하는 과정을 나누어 보기

먼저 각 행이 조건에 맞는지 `True`와 `False`로 확인한 뒤, 그 결과로 행을 선택합니다.

### D3B-02-19 · 측정소 조건 만들기

```python
station_mask = air['station'] == selected_station
```

```python
station_mask.head(15)
```

### D3B-02-20 · 측정소 조건의 True 개수 확인

```python
station_mask.value_counts()
```

### D3B-02-21 · PM2.5 조건 만들기

```python
high_mask = air['pm25'] >= pm25_limit
```

```python
high_mask.head(15)
```

### D3B-02-22 · PM2.5 조건의 True 개수 확인

```python
high_mask.value_counts()
```

### D3B-02-23 · 두 조건을 하나로 합치기

```python
combined_mask = station_mask & high_mask
```

```python
combined_mask.head(15)
```

### D3B-02-24 · 합친 조건으로 행 선택

```python
combined_rows = air[combined_mask]
```

```python
combined_rows
```

### D3B-02-25 · 두 측정소 이름 정하기

```python
# 목록의 이름을 station_names에서 골라 바꿔 보세요.
selected_stations = ['River-1', 'Central-1']
```

```python
selected_stations
```

### D3B-02-26 · 두 측정소의 행 찾기

```python
two_station_rows = air[air['station'].isin(selected_stations)]
```

```python
two_station_rows.head(10)
```

### D3B-02-27 · 두 측정소의 행 수 확인

```python
two_station_rows.shape
```

### D3B-02-28 · PM2.5 범위의 아래값 정하기

```python
# 숫자를 바꿔 범위를 조정하세요.
lower_limit = 25
```

```python
lower_limit
```

### D3B-02-29 · PM2.5 범위의 위값 정하기

```python
# 아래값보다 큰 숫자를 사용하세요.
upper_limit = 30
```

```python
upper_limit
```

### D3B-02-30 · 범위 안의 행 찾기

```python
range_rows = air[air['pm25'].between(lower_limit, upper_limit)]
```

```python
range_rows.head(10)
```

### D3B-02-31 · 범위 안의 행 수 확인

```python
range_rows.shape
```

### D3B-02-32 · PM2.5가 가장 높은 행 10개 찾기

```python
top_pm25_rows = air.nlargest(10, 'pm25')
```

```python
top_pm25_rows[['date', 'station', 'pm25']]
```

## 다른 자료에서 같은 조건 검색 사용하기

기상자료에서는 더운 날과 비가 온 날을, 콘크리트 자료에서는 재령과 강도 조건을 찾습니다.

### D3B-02-33 · 기상자료 읽기

```python
weather = pd.read_csv(DATA_DIR / 'weather_daily.csv')
```

```python
weather.head()
```

### D3B-02-34 · 최고기온 기준 정하기

```python
# 30을 28 또는 33으로 바꿔 보세요.
hot_day_limit = 30
```

```python
hot_day_limit
```

### D3B-02-35 · 더운 날 찾기

```python
hot_days = weather[weather['temperature_max_c'] >= hot_day_limit]
```

```python
hot_days[['date', 'temperature_max_c']].head(10)
```

### D3B-02-36 · 더운 날 수 확인

```python
hot_days.shape
```

### D3B-02-37 · 비가 온 날 찾기

```python
rainy_days = weather[weather['precipitation_mm'] > 0]
```

```python
rainy_days[['date', 'precipitation_mm']].head(10)
```

### D3B-02-38 · 비가 온 날 수 확인

```python
rainy_days.shape
```

### D3B-02-39 · 덥고 비가 온 날 찾기

```python
hot_and_rainy = weather[(weather['temperature_max_c'] >= hot_day_limit) & (weather['precipitation_mm'] > 0)]
```

```python
hot_and_rainy[['date', 'temperature_max_c', 'precipitation_mm']]
```

### D3B-02-40 · 콘크리트 강도자료 읽기

```python
concrete = pd.read_csv(DATA_DIR / 'concrete_strength.csv')
```

```python
concrete.head()
```

### D3B-02-41 · 재령 기준 정하기

```python
# 28을 7 또는 90으로 바꿔 보세요.
age_limit = 28
```

```python
age_limit
```

### D3B-02-42 · 정한 재령의 배합 찾기

```python
age_rows = concrete[concrete['age_days'] == age_limit]
```

```python
age_rows.head(10)
```

### D3B-02-43 · 강도 기준 정하기

```python
# MPa 단위입니다. 40을 30 또는 50으로 바꿔 보세요.
strength_limit = 40
```

```python
strength_limit
```

### D3B-02-44 · 재령과 강도 조건을 함께 적용

```python
age_and_strength = concrete[(concrete['age_days'] == age_limit) & (concrete['strength_mpa'] >= strength_limit)]
```

```python
age_and_strength.head(10)
```

### D3B-02-45 · 조건에 맞는 배합 수 확인

```python
age_and_strength.shape
```

### 확인 기록

`selected_station`, `pm25_limit`, `lower_limit`, `upper_limit` 중 하나만 바꾸고 변경 전후 행 수를 기록합니다.

## 정리

조건식의 결과는 각 행이 조건에 맞는지 나타내는 `True`와 `False`입니다. 그 결과를 표의 대괄호 안에 넣으면 원하는 행만 남길 수 있습니다.
