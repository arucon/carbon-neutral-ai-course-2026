# Day3_03 · 분석 전 데이터 확인

빈 값과 중복을 확인하고, 날짜 형식과 새 열을 준비합니다.

## 이 노트북에서 확인할 내용

- 빈 값과 중복 행을 찾을 수 있습니다.
- 원본을 보존한 채 정리할 수 있습니다.
- 날짜 형식을 바꾸고 새 열을 만들 수 있습니다.

## 실행 방법

위에서 아래로 코드 셀을 실행하고 바로 아래 결과를 확인합니다. 값을 만드는 셀 다음에는 그 값을 화면에 표시하는 셀이 이어집니다. 수정 안내가 있는 셀에서는 주석 아래의 값만 바꿉니다.

### D3B-03-01 · 경로 도구 불러오기

```python
from pathlib import Path
```

### D3B-03-02 · 현재 폴더 확인

```python
current_folder = Path.cwd()
```

```python
current_folder
```

### D3B-03-03 · 데이터 폴더 후보 만들기

```python
data_candidates = [
    current_folder / 'data',
    current_folder.parent / 'data',
]
```

```python
data_candidates
```

### D3B-03-04 · 데이터 폴더 선택

```python
DATA_DIR = next(folder for folder in data_candidates if folder.exists())
```

```python
DATA_DIR
```

### D3B-03-05 · pandas 불러오기

```python
import pandas as pd
```

### D3B-03-06 · CSV 읽기

```python
air = pd.read_csv(DATA_DIR / 'air_quality.csv')
```

```python
air
```

### D3B-03-07 · 빈 값 개수 확인

```python
missing_count = air.isna().sum()
```

```python
missing_count
```

### D3B-03-08 · 빈 값이 있는 행 찾기

```python
missing_rows = air[air.isna().any(axis=1)]
```

```python
missing_rows
```

### D3B-03-09 · 빈 값이 있는 행 수 확인

```python
missing_rows.shape
```

### D3B-03-10 · 완전히 같은 행 표시

```python
duplicate_flag = air.duplicated(keep=False)
```

```python
duplicate_flag.value_counts()
```

### D3B-03-11 · 완전히 같은 행 확인

```python
duplicate_rows = air[duplicate_flag]
```

```python
duplicate_rows
```

### D3B-03-12 · 원본을 복사

```python
clean_air = air.copy()
```

```python
clean_air.shape
```

### D3B-03-13 · 완전히 같은 행 하나만 남기기

```python
clean_air = clean_air.drop_duplicates()
```

```python
clean_air.shape
```

### D3B-03-14 · 핵심 값이 빈 행 제외하기

```python
clean_air = clean_air.dropna(subset=['pm25', 'temperature'])
```

```python
clean_air.shape
```

### D3B-03-15 · 정리 후 빈 값 다시 확인

```python
clean_air.isna().sum()
```

### D3B-03-16 · 날짜를 날짜 형식으로 바꾸기

```python
clean_air['date'] = pd.to_datetime(clean_air['date'])
```

```python
clean_air['date'].head()
```

### D3B-03-17 · 날짜 열의 값 종류 확인

```python
clean_air['date'].dtype
```

### D3B-03-18 · PM2.5가 높은지 표시하는 새 열 만들기

```python
# 아래 숫자를 바꾸면 True인 행 수가 달라집니다.
high_limit = 35
```

```python
high_limit
```

### D3B-03-19 · 새 열 추가하기

```python
clean_air['high_pm25'] = clean_air['pm25'] >= high_limit
```

```python
clean_air[['pm25', 'high_pm25']].head(10)
```

### D3B-03-20 · True와 False 개수 확인

```python
clean_air['high_pm25'].value_counts()
```

### D3B-03-27 · 강수량이 빈 행 찾기

```python
rainfall_missing_rows = clean_air[clean_air['rainfall'].isna()]
```

```python
rainfall_missing_rows
```

### D3B-03-28 · 강수량 연습용 표 복사

### D3B-03-29 · 강수량 빈 값을 0으로 바꾼 연습 표 만들기

이 셀은 값 0이 '비가 오지 않음'을 뜻한다고 정한 경우의 연습입니다. 실제 연구에서는 자료 설명을 먼저 확인합니다.

```python
rain_practice['rainfall'] = rain_practice['rainfall'].fillna(0)
```

```python
rain_practice['rainfall'].isna().sum()
```

### D3B-03-30 · 월을 새 열로 추가

```python
clean_air['month'] = clean_air['date'].dt.month
```

```python
clean_air[['date', 'month']].head()
```

### D3B-03-31 · 일을 새 열로 추가

```python
clean_air['day'] = clean_air['date'].dt.day
```

```python
clean_air[['date', 'day']].head()
```

### D3B-03-32 · 비가 온 날 표시

```python
clean_air['rainy_day'] = clean_air['rainfall'] > 0
```

```python
clean_air[['rainfall', 'rainy_day']].head(10)
```

### D3B-03-33 · 비가 온 날과 오지 않은 날 수 확인
