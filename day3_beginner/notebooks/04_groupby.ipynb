# Day3_04 · 이름별·월별 데이터 요약

측정소·건물 조건·월처럼 같은 이름이나 번호를 가진 행을 묶어 평균·개수·합계를 계산합니다.

## 이 노트북에서 확인할 내용

- 같은 이름이나 번호를 가진 행을 묶을 수 있습니다.
- 평균·개수·합계 중 질문에 맞는 계산을 선택할 수 있습니다.
- 대기질·건물·기상자료에 같은 요약 방법을 적용할 수 있습니다.

### D3B-04-02 · 현재 폴더 확인

### D3B-04-03 · 데이터 폴더 후보 만들기

```python
DATA_DIR = next(folder for folder in data_candidates if folder.exists())
```

```python
DATA_DIR
```

### D3B-04-05 · pandas 불러오기

```python
import pandas as pd
```

### D3B-04-06 · CSV 읽기

```python
air = pd.read_csv(DATA_DIR / 'air_quality.csv')
```

```python
air
```

### D3B-04-07 · 분석할 행 준비

```python
analysis_air = air.drop_duplicates()
```

```python
analysis_air.shape
```

### D3B-04-08 · PM2.5가 빈 행 제외

```python
analysis_air = analysis_air.dropna(subset=['pm25'])
```

```python
analysis_air.shape
```

### D3B-04-09 · 측정소 이름 확인

```python
station_names = analysis_air['station'].unique()
```

```python
station_names
```

### D3B-04-10 · 측정소별로 행 묶기

아직 숫자를 계산하지 않았습니다. 표가 측정소 이름을 기준으로 나뉜 상태만 확인합니다.

```python
station_groups = analysis_air.groupby('station')
```

```python
station_groups
```

### D3B-04-11 · 측정소마다 PM2.5 평균 계산

```python
station_mean = station_groups['pm25'].mean()
```

```python
station_mean
```

### D3B-04-12 · 평균이 큰 순서로 정렬

```python
station_mean_sorted = station_mean.sort_values(ascending=False)
```

```python
station_mean_sorted
```

### D3B-04-13 · 가장 높은 측정소 확인

```python
station_mean_sorted.head(1)
```

### D3B-04-14 · 측정소마다 관측 수 계산

```python
station_count
```

### D3B-04-15 · 구역마다 PM2.5 평균 계산

```python
district_mean = analysis_air.groupby('district')['pm25'].mean()
```

```python
district_mean
```

### D3B-04-16 · 평균과 관측 수를 한 표에 정리

```python
station_summary = analysis_air.groupby('station')['pm25'].agg(['mean', 'count'])
```

## 건축공학 자료 요약

건물 방향과 높이에 따라 난방·냉방 부하 평균이 어떻게 다른지 같은 방법으로 확인합니다.

### D3B-04-31 · 건물 에너지자료 읽기

```python
building = pd.read_csv(DATA_DIR / 'building_energy.csv')
```

```python
building.head()
```

### D3B-04-32 · 건물 방향 값 확인

```python
building['orientation'].unique()
```

### D3B-04-33 · 방향별 난방 부하 평균 계산

```python
heating_by_orientation = building.groupby('orientation')['heating_load'].mean()
```

```python
heating_by_orientation
```

### D3B-04-34 · 방향별 냉방 부하 평균 계산

```python
cooling_by_orientation = building.groupby('orientation')['cooling_load'].mean()
```

```python
cooling_by_orientation
```

### D3B-04-35 · 건물 높이 값 확인

```python
building['overall_height_m'].unique()
```

### D3B-04-36 · 높이별 난방·냉방 부하 평균 계산

```python
load_by_height = building.groupby('overall_height_m')[['heating_load', 'cooling_load']].mean()
```

```python
load_by_height
```

## 기상자료 요약

366개의 일별 값을 월별 평균기온과 월별 누적강수량으로 줄여 봅니다.

### D3B-04-37 · 기상자료 읽기

```python
weather = pd.read_csv(DATA_DIR / 'weather_daily.csv')
```

```python
weather.head()
```

### D3B-04-38 · 날짜를 날짜 형식으로 바꾸기

```python
weather['date'] = pd.to_datetime(weather['date'])
```

```python
weather['date'].head()
```

### D3B-04-39 · 월을 새 열로 추가

```python
weather['month'] = weather['date'].dt.month
```

```python
weather[['date', 'month']].head()
```

### D3B-04-40 · 월별 평균기온 계산

```python
monthly_temperature = weather.groupby('month')['temperature_mean_c'].mean()
```

```python
monthly_temperature
```

### D3B-04-41 · 월별 누적강수량 계산

```python
monthly_precipitation = weather.groupby('month')['precipitation_mm'].sum()
```

```python
monthly_precipitation
```

### D3B-04-42 · 월별 기상 요약표 만들기

```python
monthly_weather = pd.DataFrame({
    'temperature_mean_c': monthly_temperature,
    'precipitation_total_mm': monthly_precipitation,
})
```

```python
monthly_weather
```

## 정리

여러 행을 같은 이름끼리 묶고 평균이나 개수를 계산했습니다. pandas에서는 이 작업을 **groupby(그룹바이)**라고 부릅니다.
