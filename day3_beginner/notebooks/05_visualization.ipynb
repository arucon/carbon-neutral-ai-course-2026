### D3B-05-01 · 경로 도구 불러오기

```python
from pathlib import Path
```

### D3B-05-02 · 현재 폴더 확인

```python
current_folder = Path.cwd()
```

```python
current_folder
```

### D3B-05-03 · 데이터 폴더 후보 만들기

```python
data_candidates = [
    current_folder / 'data',
    current_folder.parent / 'data',
]
```

```python
data_candidates
```

### D3B-05-04 · 데이터 폴더 선택

```python
DATA_DIR = next(folder for folder in data_candidates if folder.exists())
```

```python
DATA_DIR
```

### D3B-05-05 · pandas 불러오기

```python
import pandas as pd
```

### D3B-05-06 · CSV 읽기

```python
air = pd.read_csv(DATA_DIR / 'air_quality.csv')
```

```python
air
```

### D3B-05-07 · 그림 도구 불러오기

```python
import matplotlib.pyplot as plt
```

### D3B-05-08 · Seaborn 불러오기

```python
import seaborn as sns
```

### D3B-05-09 · 중복 행 제외

```python
plot_air = air.drop_duplicates()
```

```python
plot_air.shape
```

### D3B-05-10 · 그림에 필요한 값이 빈 행 제외

```python
plot_air = plot_air.dropna(subset=['pm25', 'temperature'])
```

```python
plot_air.shape
```

## 1. 막대그래프

**질문:** 측정소별 평균 PM2.5는 어디가 높고 낮은가?

**필요한 데이터:** 측정소 이름과 측정소별 평균

**읽는 방법:** 막대가 길수록 평균값이 큽니다.

### D3B-05-11 · 막대그래프용 표 만들기

```python
bar_data = plot_air.groupby('station', as_index=False)['pm25'].mean()
```

```python
bar_data
```

### D3B-05-12 · 막대그래프 그리기

```python
sns.barplot(data=bar_data, x='station', y='pm25')
plt.xticks(rotation=45)
plt.title('Average PM2.5 by station')
plt.show()
```

## 2. 선그래프

**질문:** 날짜가 지나면서 평균 PM2.5가 어떻게 달라지는가?

**필요한 데이터:** 날짜와 날짜별 평균

**읽는 방법:** 선이 올라가는 구간과 내려가는 구간을 찾습니다.

### D3B-05-13 · 날짜를 날짜 형식으로 바꾸기

```python
plot_air['date'] = pd.to_datetime(plot_air['date'])
```

```python
plot_air['date'].head()
```

### D3B-05-14 · 선그래프용 표 만들기

```python
line_data = plot_air.groupby('date', as_index=False)['pm25'].mean()
```

```python
line_data.head()
```

### D3B-05-15 · 선그래프 그리기

```python
sns.lineplot(data=line_data, x='date', y='pm25')
plt.title('Daily average PM2.5')
plt.show()
```

## 3. 산점도

**질문:** 기온이 다른 날에는 PM2.5도 함께 달라지는가?

**필요한 데이터:** 같은 행의 기온과 PM2.5

**읽는 방법:** 점들이 어느 방향으로 모이는지, 멀리 떨어진 점이 있는지 확인합니다.

### D3B-05-16 · 산점도 그리기

```python
sns.scatterplot(data=plot_air, x='temperature', y='pm25')
plt.title('Temperature and PM2.5')
plt.show()
```

### D3B-05-17 · 확인할 측정소 정하기

```python
# River-1을 Central-1 또는 Park-1로 바꿔 보세요.
selected_station = 'River-1'
```

```python
selected_station
```

### D3B-05-18 · 한 측정소의 데이터 선택

```python
station_plot_data = plot_air[plot_air['station'] == selected_station]
```

```python
station_plot_data.head()
```

### D3B-05-19 · 선택한 측정소의 산점도 그리기

```python
sns.scatterplot(data=station_plot_data, x='temperature', y='pm25')
plt.title(selected_station)
plt.show()
```

## 4. 분포를 확인하는 그래프

**질문:** PM2.5 값은 어느 범위에 많이 모여 있는가?

막대 하나가 개별 측정소가 아니라 값의 구간을 뜻합니다.

### D3B-05-20 · PM2.5 히스토그램 그리기

```python
sns.histplot(data=plot_air, x='pm25', bins=15)
plt.title('Distribution of PM2.5')
plt.show()
```

### D3B-05-21 · 구간 수 정하기

```python
# 15를 8 또는 25로 바꿔 모양을 비교하세요.
bin_count = 15
```

```python
bin_count
```

### D3B-05-22 · 바꾼 구간 수로 다시 그리기

```python
sns.histplot(data=plot_air, x='pm25', bins=bin_count)
plt.title('Distribution of PM2.5')
plt.show()
```

## 5. 구역별 값의 범위

**질문:** 구역마다 PM2.5의 중앙 위치와 퍼진 정도가 어떻게 다른가?

상자 안의 선은 가운데 값을, 상자와 선의 길이는 값이 퍼진 범위를 보여 줍니다.

### D3B-05-23 · 구역별 상자그림 그리기

```python
sns.boxplot(data=plot_air, x='district', y='pm25')
plt.xticks(rotation=45)
plt.title('PM2.5 range by district')
plt.show()
```

### D3B-05-24 · 구역별 막대그래프용 표 만들기

```python
district_bar_data = plot_air.groupby('district', as_index=False)['pm25'].mean()
```

```python
district_bar_data
```

### D3B-05-25 · 구역별 평균 막대그래프 그리기

```python
sns.barplot(data=district_bar_data, x='district', y='pm25')
plt.xticks(rotation=45)
plt.title('Average PM2.5 by district')
plt.show()
```

### D3B-05-26 · 선택한 측정소의 날짜순 표 만들기

```python
station_line_data = station_plot_data.sort_values('date')
```

```python
station_line_data[['date', 'pm25']].head()
```

### D3B-05-27 · 한 측정소의 날짜별 선그래프 그리기

```python
sns.lineplot(data=station_line_data, x='date', y='pm25')
plt.title(selected_station)
plt.show()
```

### D3B-05-28 · 강수량이 있는 행 준비

```python
rain_plot_data = plot_air.dropna(subset=['rainfall'])
```

```python
rain_plot_data.shape
```

### D3B-05-29 · 강수량과 PM2.5 산점도 그리기

```python
sns.scatterplot(data=rain_plot_data, x='rainfall', y='pm25')
plt.title('Rainfall and PM2.5')
plt.show()
```

## 6. 울산 기상자료

평균기온은 선그래프로, 강수량은 막대그래프로 확인합니다.

### D3B-05-30 · 기상자료 읽기

```python
weather = pd.read_csv(DATA_DIR / 'weather_daily.csv')
```

```python
weather.head()
```

### D3B-05-31 · 기상 날짜를 날짜 형식으로 바꾸기

```python
weather['date'] = pd.to_datetime(weather['date'])
```

```python
weather['date'].head()
```

### D3B-05-32 · 일별 평균기온 선그래프 그리기

```python
sns.lineplot(data=weather, x='date', y='temperature_mean_c')
plt.title('Ulsan daily mean temperature in 2024')
plt.show()
```

### D3B-05-33 · 월을 새 열로 추가

```python
weather['month'] = weather['date'].dt.month
```

```python
weather[['date', 'month']].head()
```

### D3B-05-34 · 월별 누적강수량 표 만들기

```python
monthly_rain = weather.groupby('month', as_index=False)['precipitation_mm'].sum()
```

```python
monthly_rain
```

### D3B-05-35 · 월별 누적강수량 막대그래프 그리기

```python
sns.barplot(data=monthly_rain, x='month', y='precipitation_mm')
plt.title('Ulsan monthly precipitation in 2024')
plt.show()
```
